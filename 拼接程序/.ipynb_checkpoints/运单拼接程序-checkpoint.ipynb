```python
import os 
import pandas as pd

# 获取当前文件绝对路径
path = os.path.abspath('.')
# source_path
sp = os.path.join(path,'运单源')
```

```python
frames =[]
for root, dirs, files in os.walk(sp):
    for file in files:
        if os.path.splitext(file)[1] == '.csv':
            fp = os.path.join(root,file)
            df = pd.read_csv(fp) # excel->dataframe
            frames.append(df)
```

```python
# 多个df合并为一个
final_df = pd.concat(frames)
```

```python
#写入2.csv中
new_file_name = os.path.join(path,'2.csv')
final_df.to_csv(new_file_name, index=False)
```
